## Tokenização em sentenças, palavras e expressões regulares

Token é um pedaço de um todo, então:

uma palavra é um token em uma sentença;
uma sentença é um token em um parágrafo.

### Tokenização de sentenças

```python
from nltk.tokenize import sent_tokenize

sentence_tokenized = sent_tokenize(dataset, language='portuguese')
print(sentence_tokenized)
```

```python
len(sentence_tokenized)
```

## Normalização de texto

```python
import unicodedata 

def normalize_string(string):
    if isinstance(string, str):
        nfkd_form = unicodedata.normalize('NFKD', string.lower())
        return nfkd_form.encode('ASCII', 'ignore').decode('utf-8')
    
print(normalize_string(sentence_tokenized[0]))
```

### Tokenização em Português

Por padrão, o NLTK usa um modelo treinado em inglês. Para saber onde termina uma frase, ele precisa conhecer as abreviações da língua: em inglês, "Sra." e "Ltda." parecem fim de frase.

Por isso passamos `language='portuguese'` para o `sent_tokenize`. Compare:

```python
exemplo = 'A Sra. Souza é dona da Construtora Alfa Ltda. Ela não quis comentar.'

print(sent_tokenize(exemplo))
print(sent_tokenize(exemplo, language='portuguese'))
```

### Tokenização em palavras

Para a primeira sentença do texto.

```python
from nltk.tokenize import word_tokenize

first_sentence_word_tokenized = word_tokenize(sentence_tokenized[0], language='portuguese')
print(first_sentence_word_tokenized)
```

### Tokenização com expressão regular

Para pegar apenas palavras em um texto, sem a pontuação.

O padrão abaixo pega sequências de letras e números (`\w+`) e mantém juntas as partes ligadas por hífen ou apóstrofo. Sem isso, "quinta-feira" viraria duas palavras, "quinta" e "feira", e "porta-voz" viraria "porta" e "voz", o que atrapalha a contagem de palavras em português.

```python
from nltk.tokenize import RegexpTokenizer
tokenizer = RegexpTokenizer(r"\w+(?:[-']\w+)*")

first_sentence_word_tokenized_without_punctuation = tokenizer.tokenize(sentence_tokenized[0])
print(first_sentence_word_tokenized_without_punctuation)
```

## Filtrando stopwords!

Stopwords são palavras que geralmente não contribuem para o significado de uma sentença.

A lista do NLTK está toda em minúsculas. Por isso comparamos `word.lower()` com a lista: sem isso, "Um" no começo da frase passaria pelo filtro, porque "Um" e "um" são textos diferentes para o Python.

```python
from nltk.corpus import stopwords

portuguese_stops = set(stopwords.words('portuguese'))
words = first_sentence_word_tokenized_without_punctuation

words_without_stop = [word for word in words if word.lower() not in portuguese_stops]
print(words_without_stop)
```

## Stemming

Stemming é a técnica que remove os afixos das palavras, deixando apenas seu radical. Para o português, o NLTK tem o `RSLPStemmer`.

```python
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
stem_acidente = stemmer.stem(words_without_stop[0]) #acidente
print(stem_acidente)
```
